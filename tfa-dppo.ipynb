### Reproducing the model from the pytorch version

With the lack of existing implementations in tensorflow agents to use as reference,
an attempt was made to reproduce the model from the source code which was written in pytorch.

The various imports and functions were changed to their equivalent ones in tensorflow.
The relevant classes and pieces of code were extracted:




````python
# Adapted from: https://github.com/irom-princeton/dppo/blob/main/model/common/mlp.py

import tensorflow as tf

def get_layers_for_ResidualMLP(
        dim_list,
        activation_type="Mish",
        out_activation_type="Identity",
        use_layernorm=False,
        use_layernorm_final=False,
        dropout=0,
):
    hidden_dim = dim_list[1]
    num_hidden_layers = len(dim_list) - 3
    assert num_hidden_layers % 2 == 0
    layers = [tf.keras.layers.Dense(hidden_dim, input_shape=(dim_list[0],))]
    layers.extend(
        [
            TwoLayerPreActivationResNetLinear(
                units=hidden_dim,
                dropout_rate=dropout,
                use_normalization=use_layernorm,
            )
            for _ in range(1, num_hidden_layers, 2)
        ]
    )
    layers.append(tf.keras.layers.Dense(dim_list[-1]))
    if use_layernorm_final:
        layers.append(tf.keras.layers.LayerNormalization())
    if out_activation_type == "Identity":
        layers.append(tf.keras.layers.Activation('linear'))
    else:
        layers.append(tf.keras.layers.Activation(out_activation_type))

    return layers



class TwoLayerPreActivationResNetLinear(tf.keras.layers.Layer):
    def __init__(self, units, dropout_rate=0.0, use_normalization=False):
        super(TwoLayerPreActivationResNetLinear, self).__init__()
        self.units = units
        self.dropout_rate = dropout_rate
        self.use_normalization = use_normalization

        # Define the layers
        self.dense1 = tf.keras.layers.Dense(units, activation=None, use_bias=False)
        self.dense2 = tf.keras.layers.Dense(units, activation=None, use_bias=False)

        if self.use_normalization:
            self.norm1 = tf.keras.layers.BatchNormalization()
            self.norm2 = tf.keras.layers.BatchNormalization()

        if self.dropout_rate > 0:
            self.dropout = tf.keras.layers.Dropout(dropout_rate)

    def call(self, inputs, training=False):
        x = inputs
        # Pre-activation (Batch Normalization + Activation) before the first Dense layer
        if self.use_normalization:
            x = self.norm1(x, training=training)
        x = tf.keras.activations.relu(x)

        # First layer
        x = self.dense1(x)
        if self.dropout_rate > 0:
            x = self.dropout(x, training=training)

        # Residual connection with pre-activation for the second layer
        residual = x
        if self.use_normalization:
            x = self.norm2(x, training=training)
        x = tf.keras.activations.relu(x)

        # Second layer
        x = self.dense2(x)

        # Add the residual connection
        x += residual

        return x
````

````python
# Adapted from: https://github.com/irom-princeton/dppo/blob/main/model/diffusion/mlp_diffusion.py#L174

import tensorflow as tf

# TensorFlow doesn't have Mish by default
def mish(x):
    return x * tf.math.tanh(tf.math.softplus(x))


class DiffusionMLP(tf.keras.Model):

    def __init__(
        self,
        action_dim,
        horizon_steps,
        cond_dim,
        time_dim=16,
        mlp_dims=[256, 256],
        cond_mlp_dims=None,
        activation_type="Mish",
        out_activation_type="Identity",
        use_layernorm=False,
        residual_style=False,
    ):
        super(DiffusionMLP, self).__init__()
        output_dim = action_dim * horizon_steps
        self.time_embedding = tf.keras.Sequential([
            SinusoidalPosEmb(time_dim),
            tf.keras.layers.Dense(time_dim * 2),
            tf.keras.layers.Activation(mish),
            tf.keras.layers.Dense(time_dim),
        ])
        input_dim = time_dim + action_dim * horizon_steps + cond_dim
        self.mlp_mean = tf.keras.Sequential(get_layers_for_ResidualMLP(
            [input_dim] + mlp_dims + [output_dim],
            activation_type=activation_type,
            out_activation_type=out_activation_type,
            use_layernorm=use_layernorm,
        ))
        self.time_dim = time_dim

    # NOT CORRECT SIGNATURE!
    def forward(
        self,
        x,
        time,
        cond,
        **kwargs,
    ):
        """
        x: (B, Ta, Da)
        time: (B,) or int, diffusion step
        cond: dict with key state/rgb; more recent obs at the end
            state: (B, To, Do)
        """
        B, Ta, Da = x.shape

        # flatten chunk
        x = tf.reshape(x, [B, -1])

        # flatten history
        state = tf.reshape(cond["state"], [B, -1])

        # obs encoder
        if hasattr(self, "cond_mlp"):
            state = self.cond_mlp(state)

        # append time and cond
        time = tf.reshape(time, [B, 1])
        time_emb = tf.reshape(self.time_embedding(time), [B, self.time_dim])
        x = tf.concat([x, time_emb, state], axis=-1)

        # mlp head
        out = self.mlp_mean(x)
        return out.view(B, Ta, Da)



class SinusoidalPosEmb(tf.Module):

    def __init__(self, dim):
        super().__init__()
        self.dim = dim

    def __call__(self, x):
        half_dim = self.dim // 2
        emb = tf.math.log(tf.constant(10000, dtype=tf.float32)) / (half_dim - 1)
        emb = tf.exp(tf.range(half_dim, dtype=tf.float32) * -emb)
        emb = tf.expand_dims(x, -1) * tf.expand_dims(emb, 0)
        emb = tf.concat([tf.sin(emb), tf.cos(emb)], axis=-1)
        return emb

````

### Incompatible Model declaration

Unfortunately, the naive translation from pytorch to tensorflow did not work.

The model, as declared above, is a `tf.keras.Model`, which expects to be given a single observation `x` and a time step `time`, in addition to the state of the network.

In contrast, the tf-agent setup provides `observations` according to the environment's action spec.

### Naive adapter

A naive adapter class, which implements the `tf-agents.networks.network.Network` interface, was created and tried.
However, it was not successful.

```python
from tf_agents.networks import network

class DiffusionActorNetwork(network.Network):
    def __init__(self, input_tensor_spec, mlp, state_spec=(),
                 name='DiffusionActorNetwork'):
        super(DiffusionActorNetwork, self).__init__(
            input_tensor_spec=input_tensor_spec,
            state_spec=state_spec,
            name=name)

        self.mlp = mlp

    def call(self, observations, step_type, network_state, training=False):
        # Unpack the input observations
        observation = observations['observation']
        time = observations['time']

        # Forward pass through the DiffusionMLP
        actions = self.mlp.forward(observation, time, {'state': network_state})

        return actions, network_state
```

The problem with the adapter class was that the `observations` argument is actually just the environment's action spec,
and it does not contain the time component separately by default.

### Replicate the TwoLayer MDP

In the paper, they describe how the model can be viewed as a two-layer Markov Decision Process.

The outer "environment" layer corresponds to actual interactions with the environment.

Whereas the inner layer represents multiple internal denoising steps that happen for each single action.

In that setup, the internal denoising steps do not correspond to actual actions and do not get rewards.

#### Implementation goal

To replicate that aspect, the rewards for internal steps should be forced to be `0`,
 and only the actual step that corresponds to taking an actition in the environment should have an actual reward.

## Metrics and Evaluation

The most common metric used to evaluate a policy is the average return. The return is the sum of rewards obtained while running a policy in an environment for an episode, and we usually average this over a few episodes. We can compute the average return metric as follows.


````python
#@test {"skip": true}
def compute_avg_return(environment, policy, num_episodes=10):

  total_return = 0.0
  for _ in range(num_episodes):

    time_step = environment.reset()
    episode_return = 0.0

    while not time_step.is_last():
      action_step = policy.action(time_step)
      time_step = environment.step(action_step.action)
      episode_return += time_step.reward
    total_return += episode_return

  avg_return = total_return / num_episodes
  return avg_return.numpy()[0]


random_policy = random_tf_policy.RandomTFPolicy(train_env.time_step_spec(),
                                                train_env.action_spec())

compute_avg_return(eval_env, random_policy, num_eval_episodes)

# Please also see the metrics module for standard implementations of different
# metrics.
````

## Data Collection

As in the DQN tutorial, set up the replay buffer and the initial data collection with the random policy.

````python
#@test {"skip": true}
replay_buffer = tf_uniform_replay_buffer.TFUniformReplayBuffer(
    data_spec=agent.collect_data_spec,
    batch_size=train_env.batch_size,
    max_length=replay_buffer_capacity)

def collect_step(environment, policy):
  time_step = environment.current_time_step()
  action_step = policy.action(time_step)
  next_time_step = environment.step(action_step.action)
  traj = trajectory.from_transition(time_step, action_step, next_time_step)

  # Add trajectory to the replay buffer
  replay_buffer.add_batch(traj)

for _ in range(initial_collect_steps):
  collect_step(train_env, random_policy)

# This loop is so common in RL, that we provide standard implementations of
# these. For more details see the drivers module.

# Dataset generates trajectories with shape [BxTx...] where
# T = n_step_update + 1.
dataset = replay_buffer.as_dataset(
    num_parallel_calls=3, sample_batch_size=batch_size,
    num_steps=n_step_update + 1).prefetch(3)

iterator = iter(dataset)
````

## Training the agent

The training loop involves both collecting data from the environment and optimizing the agent's networks. Along the way, we will occasionally evaluate the agent's policy to see how we are doing.

The following will take ~7 minutes to run.

````python
#@test {"skip": true}
try:
  %%time
except:
  pass

# (Optional) Optimize by wrapping some of the code in a graph using TF function.
agent.train = common.function(agent.train)

# Reset the train step
agent.train_step_counter.assign(0)

# Evaluate the agent's policy once before training.
avg_return = compute_avg_return(eval_env, agent.policy, num_eval_episodes)
returns = [avg_return]

for _ in range(num_iterations):

  # Collect a few steps using collect_policy and save to the replay buffer.
  for _ in range(collect_steps_per_iteration):
    collect_step(train_env, agent.collect_policy)

  # Sample a batch of data from the buffer and update the agent's network.
  experience, unused_info = next(iterator)
  train_loss = agent.train(experience)

  step = agent.train_step_counter.numpy()

  if step % log_interval == 0:
    print('step = {0}: loss = {1}'.format(step, train_loss.loss))

  if step % eval_interval == 0:
    avg_return = compute_avg_return(eval_env, agent.policy, num_eval_episodes)
    print('step = {0}: Average Return = {1:.2f}'.format(step, avg_return))
    returns.append(avg_return)
````
